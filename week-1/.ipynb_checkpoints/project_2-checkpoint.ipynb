```python
#input phase
p= float(input("enter principal amount: "))
r= float(input("enter interest rate: "))
n= float(input("no. of times compounded a year: "))
t= float(input("time period of the investment: "))

#processing phase
tot_amt= p*(1+((r/100)/n))**(n*t)
comp_interest= tot_amt-p

#output phase
print(f"\n-------RESULTS-------")
print(f"Total Amount: {tot_amt:.2f}")
print(f"Compounded Interest: {comp_interest:.2f}")
```
