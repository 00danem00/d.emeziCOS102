```python
#input phase
principal= float(input("Enter the principal amount (P): "))
rate= float(input("Enter rate (R) in %: "))
time= float(input("Enter the time period (T) in years: "))

#processing phase
#formula: A = P(1+(R/100)*T)
tot_amt= principal*(1+(rate/100)*time)
simple_interest= tot_amt-principal

#output phase
print(f"\n-------RESULTS-------")
print(f"Total Amount: {tot_amt:.3f}")
print(f"Interest Earned: {simple_interest:.2f}")
```
