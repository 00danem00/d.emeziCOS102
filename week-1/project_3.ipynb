```python
#input phase
pmt= float(input("enter periodic payment: "))
r= float(input("enter annual interest rate: "))
n= float(input("compounding periods per yr: "))
t= float(input("number of years: "))

#processing phase
i= r/n
periods= n*t
amount= pmt*(((1+i)**periods-1)/i)

#output phase
print(f"\n-------RESULTS-------")
print(f"The futue value of the annuity is: {amount:.2f}")
```
