```python
import tkinter as tk
from tkinter import messagebox
import csv

#Store all employee data inside this list
employees = []

#load CSV file
def load_data():
    try:
        #open the csv file
        with open('GIG-logistics.csv', 'r') as file:
            reader = csv.DictReader(file)

            #loop through each row in the csv
            for row in reader:
                employees.append(row)

    except FileNotFoundError:
        messagebox.showerror("Error", "CSV file not found")


#verify employee
def verify_employee():
    #get user input
    name = name_entry.get()
    department = dept_entry.get()

    if not name or not department:
        messagebox.showwarning("Input Error", "Please fill all fields")
        return

    #convert to lowercase (to avoid issues)
    name = name.lower().strip()
    department = department.lower().strip()

    found = False
    department_members = []

    #loop through employees
    for emp in employees:
        emp_name = emp['Name'].lower().strip()
        emp_dept = emp['Department'].lower().strip()

        #collect everyone in same dapartment
        if emp_dept == department:
            department_members.append(emp['Name'])

        #check if employee exists
        if emp_name == name and emp_dept == department:
            found = True

    #show results
    if found:
        message = "welcome, " +name.title() + "!\n\n"
        message += "other members in your department:\n"

        for member in department_members:
            message += member + "\n"

        messagebox.showinfo("success", message)

    else:
        messagebox.showwarning("not found", "employee doesnt exist")


#build the GUI
root = tk.Tk()
root.title("Employee verification system")
root.geometry("400x250")

#name label and input
name_label = tk.Label(root, text="enter name: ")
name_label.pack()
name_entry = tk.Entry(root)
name_entry.pack()

#department label and input
dept_label = tk.Label(root, text ="enter department: ")
dept_label.pack()
dept_entry = tk.Entry(root)
dept_entry.pack()

#button to verify
verify_button = tk.Button(root, text="verify", command=verify_employee)
verify_button.pack(pady=10)

#load data when program starts
load_data()

#run the GUI
root.mainloop()
```
