```python
print("Welcome to our Annual Tax Revenue(ATR) calculator")
age = int(input("Input ur age: "))
xp = int(input("Input work experience in years: "))

if xp>25 and age>=55:
    print("Your ATR is: N5,600,000")
elif xp>20 and age>=45:
    print("Your ATR is: N4,480,000")
elif xp>10 and age>=35:
    print("Your ATR is: N1,500,000")
elif xp<10 and age<35:
    print("Your ATR is: N550,000")
else: 
    print("N0.00")
```
