```python
import numpy as np


print("1. Quadratic Equation")
print("2. Cubic Equation")
print("3. Quartic Equation")

choice = int(input("Choose the equation type (1, 2, or 3): "))

# Quadratic
if choice == 1:
    print("\nEnter coefficients for the equation")

    a = float(input("Enter a: "))
    b = float(input("Enter b: "))
    c = float(input("Enter c: "))

    coefficients = [a, b, c]
    roots = np.roots(coefficients)

    print("\nRoots of the equation are: ", roots)

# Cubic
elif choice == 2:
    print("\nEnter coefficients for the equation")

    a = float(input("Enter a: "))
    b = float(input("Enter b: "))
    c = float(input("Enter c: "))
    d = float(input("Enter d: "))

    coefficients = [a, b, c, d]
    roots = np.roots(coefficients)

    print("\nRoots of the equation are:", roots)

# Quartic
elif choice == 3:
    print("\nEnter coefficients for the equation")

    a = float(input("Enter a: "))
    b = float(input("Enter b: "))
    c = float(input("Enter c: "))
    d = float(input("Enter d: "))
    e = float(input("Enter e: "))

    coefficients = [a, b, c, d, e]
    roots = np.roots(coefficients)

    print("\nRoots of the equation are:", roots)

else:
    print("Invalid choice")
```
